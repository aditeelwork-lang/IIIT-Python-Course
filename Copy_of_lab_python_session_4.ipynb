# 📖 WHAT ARE LISTS?

Lists store **multiple items** in one variable!

## Creating Lists:

```python
# Empty list
my_list = []

# List with items
fruits = ["apple", "banana", "orange"]
numbers = [1, 2, 3, 4, 5]
mixed = ["hello", 42, True, 3.14]
```

## Why Use Lists?

Instead of:
```python
fruit1 = "apple"
fruit2 = "banana"
fruit3 = "orange"
```

Use:
```python
fruits = ["apple", "banana", "orange"]
```

Much better! 🎉

````python
# 👀 DEMO: Lists in Action

# Create a list
colors = ["red", "green", "blue"]

print("My list:", colors)
print("Number of items:", len(colors))
print()

# Access items
print("First color:", colors[0])
print("Second color:", colors[1])
print("Last color:", colors[-1])
````

```python

fruits.insert(1, "grape")

# 📖 REMOVING FROM LISTS

```python
fruits = ["apple", "banana", "orange", "grape"]

# Remove by value
fruits.remove("banana")
# Now: ["apple", "orange", "grape"]

# Remove by index
fruits.pop(0)
# Now: ["orange", "grape"]

# Remove last item
fruits.pop()
# Now: ["orange"]

# Clear entire list
fruits.clear()
# Now: []
```

# 📖 ACCESSING ITEMS BY INDEX

Lists use **zero-based indexing**:

```python
fruits = ["apple", "banana", "orange", "grape"]
#          0        1         2         3

print(fruits[0])   # apple
print(fruits[2])   # orange
print(fruits[-1])  # grape (last item)
print(fruits[-2])  # orange (second to last)
```

# 🏆 FINAL PROJECT: Shopping List Manager!

Build an interactive shopping list app with these features:

1. Show menu with options
2. Add items to list
3. View all items
4. Remove items
5. Exit

## Reward: +30 XP

````python
print("🛒 SHOPPING LIST MANAGER 🛒")
print("="*50)

shopping_list = []

while True:
    print("\n1. Add item")
    print("2. View list")
    print("3. Remove item")
    print("4. Exit")

    choice = input("\nChoice: ")

    if choice == "1":
        item = input("Item to add: ")
        shopping_list.append(item)
        print(f"✅ Added {item}!")

    elif choice == "2":
        if shopping_list:
            print("\n📋 Your Shopping List:")
            for i, item in enumerate(shopping_list, 1):
                print(f"  {i}. {item}")
        else:
            print("📋 List is empty!")

    elif choice == "3":
        if shopping_list:
            print("\n📋 Your items:")

            for i, item in enumerate(shopping_list, 1):
                print(f"  {i}. {item}")

            num = int(input("Enter item number to remove: "))
            removed = shopping_list.pop(num - 1)

            print(f"❌ Removed {removed}")

        else:
            print("📋 List is empty!")

    elif choice == "4":
        print("👋 Goodbye!")
        break

    else:
        print("❌ Invalid choice!")
````
