# 📖 FOR LOOPS

For loops repeat code a **specific number of times**.

```python
for i in range(5):
    print("Hello!")
```

This prints "Hello!" 5 times!

## Syntax:
```python
for variable in range(count):
    # code to repeat
```

````python
# 👀 DEMO: For Loop

print("Counting to 5:")
for i in range(1, 6):
    print(f"Number: {i}")

print("\n✨ The loop repeated 5 times!")
````

# 📖 RANGE FUNCTION

`range()` generates numbers:

```python
range(5)        # 0, 1, 2, 3, 4
range(1, 6)     # 1, 2, 3, 4, 5
range(0, 10, 2) # 0, 2, 4, 6, 8
```

Syntax:
- `range(stop)` - from 0 to stop-1
- `range(start, stop)` - from start to stop-1
- `range(start, stop, step)` - with custom step

# 📖 WHILE LOOPS

While loops repeat **while a condition is true**.

```python
count = 0
while count < 5:
    print(count)
    count += 1
```

⚠️ **Important:** Make sure the condition eventually becomes False, or you'll have an infinite loop!
