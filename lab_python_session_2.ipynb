# 📖 WHAT ARE IF STATEMENTS?

If statements let your program **make decisions**!

## Real Life Example:
IF it's raining:
Take an umbrella
ELSE:
Leave umbrella at home

## Python Example:

```python
weather = "rainy"

if weather == "rainy":
    print("Take an umbrella! ☔")
else:
    print("No umbrella needed! ☀️")
```

---

## 🔑 Key Concepts:

- `if` checks a condition
- `:` (colon) starts the if block
- **Indent** the code that runs when true
- `else` runs when condition is false

````python
# 👀 DEMO: If Statement in Action

age = 18

if age >= 18:
    print("✅ You're an adult!")
    print("✅ You can vote!")
else:
    print("❌ You're not an adult yet")
    print("❌ Wait a few more years")

print("\n✨ The program decided based on age!")
````

# 📖 COMPARISON OPERATORS

These operators compare values:

| Operator | Meaning | Example |
|----------|---------|---------|
| `==` | Equal to | `5 == 5` → True |
| `!=` | Not equal | `5 != 3` → True |
| `>` | Greater than | `10 > 5` → True |
| `<` | Less than | `3 < 7` → True |
| `>=` | Greater or equal | `5 >= 5` → True |
| `<=` | Less or equal | `4 <= 6` → True |

⚠️ **Important:** Use `==` to compare, `=` to assign!

```python
x = 5      # Assign 5 to x
x == 5     # Check if x equals 5
```

# 📖 ELIF (ELSE IF)

When you need to check **multiple conditions**:

```python
score = 85

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
else:
    print("Grade: F")
```

Python checks conditions **top to bottom** and stops at the first True one.

````python
temperature = 45  # Change this to test

# TODO: Write if/elif/else chain
# ADD IF
if temperature >=50:
    print("It's hot!")
# ADD ELIF
elif temperature >=40:
    print("Nice weather!")
# ADD ELIF
elif temperature >=30:
    print("It's cold!")
# ADD ELSE
    print("Freezing!")
````
