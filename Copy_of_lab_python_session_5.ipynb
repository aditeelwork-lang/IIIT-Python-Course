# 📖 WHAT ARE DICTIONARIES?

Dictionaries store **key-value pairs**!

Think of a real dictionary:
- **Word** (key) → **Definition** (value)
- **Name** (key) → **Phone Number** (value)

## Creating Dictionaries:

```python
# Empty dictionary
my_dict = {}

# Dictionary with data
person = {
    "name": "Alice",
    "age": 25,
    "city": "New York"
}

# Access values
print(person["name"])   # Alice
print(person["age"])    # 25
```

# 📖 ADDING & MODIFYING

```python
person = {"name": "Alice", "age": 25}

# Add new key-value
person["city"] = "Boston"
# Now: {"name": "Alice", "age": 25, "city": "Boston"}

# Modify existing value
person["age"] = 26
# Now: {"name": "Alice", "age": 26, "city": "Boston"}

# Remove key-value
del person["city"]
# Now: {"name": "Alice", "age": 26}
```

# 📖 LOOPING THROUGH DICTIONARIES

```python
person = {"name": "Alice", "age": 25, "city": "NYC"}

# Loop through keys
for key in person:
    print(key)

# Loop through values
for value in person.values():
    print(value)

# Loop through both
for key, value in person.items():
    print(f"{key}: {value}")
```

# 🏆 FINAL PROJECT: Contact Book!

Build a contact book manager:

1. Add contacts (name → phone)
2. View all contacts
3. Search for a contact
4. Delete a contact
5. Exit

## Reward: +30 XP

````python
print("📱 CONTACT BOOK MANAGER 📱")
print("="*50)

contacts = {}

while True:
    print("\n1. Add contact")
    print("2. View all")
    print("3. Search contact")
    print("4. Delete contact")
    print("5. Exit")

    choice = input("\nChoice: ")

    if choice == "1":
        name = input("Name: ")
        phone = input("Phone: ")
        # Add here
        print(f"✅ Added {name}!")

    elif choice == "2":
        if contacts:
            print("\n📱 All Contacts:")
            for # Add here:
                print(f"  {name}: {phone}")
        else:
          print("📱 No contacts yet!")

    elif choice == "3":
        name = input("Name to search: ")
        if name in contacts:
            print(f"📞 {name}: {contacts[name]}")
        else:
            print(f"❌ {name} not found!")

    elif choice == "4":
        name = input("Name to delete: ")
        if name in contacts:
            # Add here
            print(f"❌ Deleted {name}")
        else:
            print(f"❌ {name} not found!")

    elif choice == "5":
        print("👋 Goodbye!")
        break
    else:
        print("❌ Invalid choice!")
````
